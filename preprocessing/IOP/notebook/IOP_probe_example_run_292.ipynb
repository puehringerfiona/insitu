# IOP Probe Processing Example

This notebook demonstrates the function-based IOP probe pipeline using the example profile:

- Folder: `C:\\Users\\airani\\Documents\\supervisions\\Fiona\\IOP_data_Fiona\\extracted\\20240618_BIE_IOP_extracted`
- Run: `292`

It generates synchronized CSV files, high-resolution PNG plots, and a single HTML dashboard.

```python
from pathlib import Path
import sys

# Make the local iop_probe package importable when this notebook is opened from the notebook folder.
PROJECT_DIR = Path.cwd().parent if Path.cwd().name.lower() == "notebook" else Path.cwd()
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from iop_probe import run_profile, run_lab

PROJECT_DIR
```

This runs the full vertical-profile workflow: ACS + VSF + CTD reading, CTD-time synchronization, down/up splitting, raw CSV export, ACS-QC CSV export, PNG plotting, and dashboard creation.

```python
input_dir = Path(r"C:\Users\airani\Documents\supervisions\Fiona\IOP_data_Fiona\extracted\20240619_WAL_IOP_extracted")
run_number = "293"
profile_output_dir = PROJECT_DIR / "notebook_output_profile_293"

profile_result = run_profile(
    input_dir=input_dir,
    run=run_number,
    output_dir=profile_output_dir,
    depth_offset_m=0.0,
)

profile_result
```

```python
def list_outputs(result):
    print(f"Output directory: {result['output_dir']}")
    print(f"Dashboard: {result['dashboard']}")
    print("\nCSV files:")
    for path in result["csv_files"]:
        print(f"  - {path.name}")
    print("\nPlot files:")
    for path in result["plot_files"]:
        print(f"  - {path.name}")
    if result["errors"]:
        print("\nWarnings:")
        for error in result["errors"]:
            print(f"  - {error}")

list_outputs(profile_result)
```

## Clean-Water Lab-Style Processing

This uses the same ACS file as a smoke test for the lab workflow. For actual clean-water measurements, change `lab_input_dir` and `lab_run_number` to the relevant extracted folder and run.

```python
lab_input_dir = input_dir
lab_run_number = run_number
lab_output_dir = PROJECT_DIR / "notebook_output_lab_292"

lab_result = run_lab(
    input_dir=lab_input_dir,
    run=lab_run_number,
    output_dir=lab_output_dir,
    trim_percent=30,
)

list_outputs(lab_result)
```
